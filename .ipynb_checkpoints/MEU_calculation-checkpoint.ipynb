```python
#possible actions
actions = [
    "Disarm_Security",
    "HVAC_Eco_Mode",
    "Activate_Siren",
    "Emergency_Dispatch"
]

states = [
    "Normal",
    "Resident_Sleeping",
    "Leaving_Home",
    "Returning_Home",
    "Unusual_Activity"
]
```

```python
#define the utility matrix
import pandas as pd

utility = pd.DataFrame(
    {
        "Normal":              [10, 8, -20, -30],
        "Resident_Sleeping":   [5, 10, -25, -30],
        "Leaving_Home":        [10, 5, 5, -20],
        "Returning_Home":      [10, 5, -20, -25],
        "Unusual_Activity":    [-10, 0, 20, 15]
    },
    index=actions
)

print(utility)
```

```python
# calculate expected utility 
state_probabilities = {
    "Normal": 0.20,
    "Resident_Sleeping": 0.40,
    "Leaving_Home": 0.10,
    "Returning_Home": 0.10,
    "Unusual_Activity": 0.20
}

expected_utility = utility.mul(
    pd.Series(state_probabilities)
).sum(axis=1)

print(expected_utility.sort_values(ascending=False))
```

```python
# Take the action with maximuk utility 
best_action = expected_utility.idxmax() # action name
best_utility = expected_utility.max()   # utility value for the action

print("Optimal action:", best_action)
print("Expected utility:", best_utility)
```
